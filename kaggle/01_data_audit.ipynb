```python
import os
import json
import pandas as pd
import numpy as np

print("Starting TrackPulse Dataset Audit...")

# File inventory
DATASET_FILES = [
    'ir_train.csv',
    'ir_test.csv',
    'ir_data_dictionary.csv',
    'etrain_delays.csv',
    'stations.json',
    'Trains schedule.csv',
    'trains.csv',
    'trains.json'
]

LEAKAGE_COLUMNS = [
    'primary_delay_cause',
    'delay_minutes',
    'actual_arrival',
    'future_delay',
    'target',
    'is_delayed'
]

print(f"Target Leakage Exclusion List: {LEAKAGE_COLUMNS}")
```

```python
def audit_dataset_structure(name, df):
    print(f"=== AUDITING DATASET: {name} ===")
    print(f"Rows: {len(df):,}, Columns: {len(df.columns)}")
    null_pct = (df.isnull().sum() / len(df) * 100).round(2)
    summary_df = pd.DataFrame({
        'dtype': df.dtypes,
        'null_pct': null_pct,
        'unique_count': df.nunique()
    })
    print(summary_df.head(10))
    print("-" * 50)
    return summary_df
```
