### SIH26028 Dynamic ETA & Multi-Train Delay Intelligence

Trains pinball loss gradient boosted quantile regressors:
- Quantile $\alpha = 0.10$ (Optimistic arrival bound P10)
- Quantile $\alpha = 0.50$ (Most likely arrival estimate P50)
- Quantile $\alpha = 0.90$ (Conservative arrival bound P90)
- Enforces monotonic non-crossing sorting constraint: $P10 \le P50 \le P90$
- Exports models: `eta_p10.txt`, `eta_p50.txt`, `eta_p90.txt`.

```python
import lightgbm as lgb
from sklearn.metrics import mean_absolute_error, mean_squared_error

def train_quantile_model(X_train, y_train, alpha):
    params = {
        'objective': 'quantile',
        'alpha': alpha,
        'metric': 'quantile',
        'learning_rate': 0.05,
        'n_estimators': 200
    }
    model = lgb.LGBMRegressor(**params)
    model.fit(X_train, y_train)
    return model

print("Quantile ETA Pipeline Initialized for alpha in [0.1, 0.5, 0.9].")
```
