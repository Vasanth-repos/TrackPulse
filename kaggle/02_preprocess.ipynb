```python
import pandas as pd
import numpy as np
import json

def normalize_train_id(val):
    if pd.isna(val):
        return ""
    s = str(val).strip()
    if s.endswith('.0'):
        s = s[:-2]
    return s.zfill(5) if len(s) < 5 and s.isdigit() else s

def normalize_station_code(val):
    if pd.isna(val):
        return ""
    return str(val).strip().upper()

print("ID Normalization Functions Initialized.")
```
