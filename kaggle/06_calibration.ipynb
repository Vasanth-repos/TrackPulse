Measures empirical coverage, sharpness, and reliability metrics across test journeys:
- Empirical 80% coverage check (verifies $y \in [P10, P90]$ matches nominal ~80% target)
- Mean Absolute Error (MAE), RMSE, and $\pm 5$ min / $\pm 10$ min arrival accuracy
- Calibration curves and reliability score calibration (0 to 100)
- Baseline benchmark comparison against static scheduled arrival and static current delay extrapolation.

```python
import numpy as np
import pandas as pd

def evaluate_interval_coverage(y_true, y_p10, y_p90):
    covered = (y_true >= y_p10) & (y_true <= y_p90)
    coverage_pct = np.mean(covered) * 100.0
    avg_width = np.mean(y_p90 - y_p10)
    print(f"Observed 80% Interval Coverage: {coverage_pct:.2f}%")
    print(f"Average Interval Width: {avg_width:.2f} minutes")
    return coverage_pct, avg_width

print("Calibration Metrics Engine Ready.")
```
